```python
import keyboard
intervalBetweenActions = 0.05

def checkPlacementValidity(whitePixelSensitivity = 5, redPixelSensitivity=5):
    im = np.array(pg.screenshot())
    mouseX, mouseY = pg.position()
    croppedIm = im[mouseY - 100: mouseY + 50, mouseX - 100 : mouseX + 100]
    if np.median((croppedIm == np.array([255, 255, 255])).all(axis=2).sum(axis=1)) > whitePixelSensitivity:
        return "Valid"
    elif np.median((croppedIm == np.array([255, 0, 0])).all(axis=2).sum(axis=1)) > redPixelSensitivity:
        return "Invalid"
    else:
        return "Unselected"


def selectAndPlaceTower(key, mouseX, mouseY):
    keyboard.press(key)
    time.sleep(intervalBetweenActions)
    keyboard.release(key)
    pg.moveTo(mouseX, mouseY)
    time.sleep(intervalBetweenActions)
    placementValidity = checkPlacementValidity()
    if placementValidity == "Valid":
        pg.click()
        pg.moveTo(960, 0)
        time.sleep(intervalBetweenActions)
        pg.click()
    return placementValidity
    
```

```python
def checkInRound(im):
    brightness = np.array(im)[-100:, -100:].mean()
    if brightness > 130 and brightness < 137:
        return True

def startRound():
    keyboard.press_and_release("space")
    time.sleep(intervalBetweenActions)
    im = pg.screenshot()
    brightness = np.array(im)[-100:, -100:].mean()
    if brightness < 130:
        keyboard.press_and_release("space")

def checkBetweenRounds(im):
    brightness = np.array(im)[-100:, -100:].mean()
    return brightness > 137

def checkDefeated(im):
    return (np.array(im)[300:400, 725:1200, 0] == 255).sum() > 25000

def checkInLevelupScreen(im):
    processedIm = np.array(im)[550:650, 750: 1150]
    levelUpWhitePixCount = (processedIm == np.array([255, 255, 255])).all(axis=2).sum()
    return (levelUpWhitePixCount > 3400) and (levelUpWhitePixCount < 3600)

def checkInUnlockUpgrade(im):
    processedIm = np.array(im)[-100:, -350:]
    return (processedIm.mean() > 86) and (processedIm.mean() < 87)
```

```python
def clickThroughLevelUpScreen():
    pg.click()
    time.sleep(2)
    pg.click()
    time.sleep(3)
```

```python
import random

# global tower placement keys
towerPlacementKeys = "qwertyzxcvbnasdfgjkl"

TOWER_X_MIN = 150
TOWER_X_MAX = 1550
TOWER_Y_MIN = 150
TOWER_Y_MAX = 950

def sampleRandomPurchase():
    towerKey = random.choice(towerPlacementKeys)
    coordinates = [
        (random.randint(TOWER_X_MIN, TOWER_X_MAX), random.randint(TOWER_Y_MIN, TOWER_Y_MAX))
        for _ in range(10)
    ]
    return towerKey, coordinates

def sampleRandomUpgrade(boughtTowers):
    towerKey = random.choice(list(boughtTowers.keys()))
    towerPlacementIdx = random.choice(range(boughtTowers[towerKey]))
    upgradeKey = random.choice([",", ".", "/"])
    return towerKey, towerPlacementIdx, upgradeKey

def generateRandomActionTape(numActions=40):
    actionTape = []
    boughtTowers = {}  # track bought towers for upgrades

    for i in range(numActions):
        # always start with buy
        actionIdentifier = "buy" if i <= 1 else random.choice(["buy", "upgrade"])

        if actionIdentifier == "buy":
            towerKey, coordinates = sampleRandomPurchase()
            if i == 0:
                towerKey = 'u'
            
            actionTape.append(("buy", (towerKey, coordinates)))
            if i != 0:
                if towerKey not in boughtTowers:
                    boughtTowers[towerKey] = 0
                boughtTowers[towerKey] += 1

        else:  # upgrade

            towerKey, towerPlacementIdx, upgradeKey = sampleRandomUpgrade(boughtTowers)
            actionTape.append(("upgrade", (upgradeKey, towerKey, towerPlacementIdx)))

    return actionTape

# example usage
tape = generateRandomActionTape(10)
for action in tape:
    print(action)
```

```python
from collections import defaultdict

def isUpgradeLogiciallyValid(upgradeKey, upgradeInfo):

    # check if already T5 in selected upgrade
    if upgradeInfo[upgradeKey] == 5:
        return False

    # check if crosspathed between other two paths already 
    crosspathedOther = True
    for key, upgradePoints in upgradeInfo.items():
        if key == upgradeKey:
            continue
        if upgradePoints == 0:
            crosspathedOther = False

    if crosspathedOther:
        return False

    # check if this at two and one of the others 3 or more
    if upgradeInfo[upgradeKey] == 2:
        for key, upgradePoints in upgradeInfo.items():
            if key == upgradeKey:
                continue
            if upgradePoints >= 3:
                return False

    return True

def runTapeUntilWaiting(tape, tapeStartIdx, towerInfoDictionary):
    waitingForMoney = False
    tapeIdx = tapeStartIdx
    while not waitingForMoney:
        currentAction = tape[tapeIdx]
        actionIdentifier, actionParameters = currentAction
        if actionIdentifier == "buy":
            towerKey, listOfCoordinates = actionParameters
            for coordinate in listOfCoordinates:
                placementValidity = selectAndPlaceTower(towerKey, *coordinate)
                if placementValidity == "Valid":

                    # (coordinates, upgradeStats) tuple
                    towerInfoDictionary[towerKey].append((coordinate, {',' : 0, "." : 0, "/" : 0}))
                    tapeIdx += 1
                    break

                elif placementValidity == "Invalid":
                    continue

                elif placementValidity == "Unselected":
                    waitingForMoney = True
                    break

                else:
                    raise Exception("SelectAndPlace Returned Something Unexpected")

            if placementValidity == "Invalid":
                tapeIdx += 1
            
            pg.moveTo(1920, 300)

        elif actionIdentifier == "upgrade":
            upgradeKey, towerKey, towerIdx = actionParameters
            towersMatchingKey = towerInfoDictionary.get(towerKey)
            if towersMatchingKey is None:
                print(f"Warning: Was not able to find tower {towerKey} to upgrade.")
                tapeIdx += 1
                continue

            elif len(towersMatchingKey) <= towerIdx:
                print(f"Warning: Idx larger than list of {towerKey} placements.")
                tapeIdx += 1
                continue

            coordinate, upgradeInfo = towersMatchingKey[towerIdx]

            if not isUpgradeLogiciallyValid(upgradeKey, upgradeInfo):
                tapeIdx += 1
                continue
            
            upgradeSuccess = attemptUpgrade(upgradeKey, *coordinate)
            if upgradeSuccess:
                upgradeInfo[upgradeKey] += 1
                tapeIdx += 1
            else:
                waitingForMoney = True
                break
        time.sleep(intervalBetweenActions)
    return towerInfoDictionary, tapeIdx
                    

def rolloutTape(tape):
    towerInfoDictionary = defaultdict(list)
    tapeIdx = 0
    rollingTape = True
    while rollingTape:
        interface.analyzeTopbar()
        topbarStats = [interface.getRound(), interface.getMoney(), interface.getHealth()]
        print(tapeIdx)
        towerInfoDictionary, tapeIdx = runTapeUntilWaiting(tape, tapeIdx, towerInfoDictionary)

        startRound()
        
        inRound = True
        while inRound:
            time.sleep(1)
            im = pg.screenshot()
            if checkInLevelupScreen(im):
                inRound = False
                clickThroughLevelUpScreen()
            elif checkDefeated(im):
                inRound = False
                rollingTape = False
            elif checkBetweenRounds(im):
                inRound = False
            elif checkInRound(im):
                pass
            else:
                print("Warning, in unknown menu")
                display(im)
    return topbarStats
        
```

```python
import random

def probabilistic_adjacent_swap(lst, actionTranspositionChance):
    result = lst.copy()
    
    for i in range(0, len(result) - 1, 2):
        if random.random() < actionTranspositionChance:
            result[i], result[i + 1] = result[i + 1], result[i]
    
    return result

def mutateCoordinates(oldCoordinateList, buyCoordinateModificationChance):
    newCoordinateList = []
    for xCoord, yCoord in oldCoordinateList:
        if random.random() < buyCoordinateModificationChance:
            newXCoord = int(xCoord + 20 * (random.random() - 0.5))
            newXCoord = max(newXCoord, TOWER_X_MIN)
            newXCoord = min(newXCoord, TOWER_X_MAX)
            newYCoord = int(yCoord + 20 * (random.random() - 0.5))
            newYCoord = max(newYCoord, TOWER_Y_MIN)
            newYCoord = max(newYCoord, TOWER_Y_MAX)
        else:
            newXCoord = xCoord
            newYCoord = yCoord
        
        newCoordinateList.append((newXCoord, newYCoord))
    return newCoordinateList

def mutateTowerKey(oldTowerKey, buyTowerReselectionChance):
    if random.random() < buyTowerReselectionChance:
        towerKey = random.choice(towerPlacementKeys)
    else:
        towerKey = oldTowerKey
    return towerKey

def mutateUpgradeKey(oldUpgradePath, upgradePathModificationChance):
    if random.random() < upgradePathModificationChance:
        upgradeKey = random.choice([",", ".", "/"])
    else:
        upgradeKey = oldUpgradePath
    return upgradeKey

def mutateUpgradeTowerSelectionKey(oldTowerKey, oldPlacementIdx, boughtTowers, upgradeTowerModificationChance):
    if random.random() < upgradeTowerModificationChance or boughtTowers[oldTowerKey] == 0:
        towerKey = random.choice([x for x in list(boughtTowers.keys()) if boughtTowers[x] > 0])
        towerPlacementIdx = random.choice(range(boughtTowers[towerKey]))
    else:
        towerKey = oldTowerKey
        towerPlacementIdx = oldPlacementIdx
    return towerKey, towerPlacementIdx

def modifyTape(tape,
               actionTypeFlipChance = 0.1,
               actionTranspositionChance = 0.1,
               buyTowerReselectionChance = 0.1,
               buyCoordinateModificationChance = 0.1,
               upgradePathModificationChance = 0.1,
               upgradeTowerModificationChance = 0.1
              ):
    
    boughtTowers = defaultdict(int)

    transposedTape = [tape[0], tape[1]]

    newTapePostTransposition = probabilistic_adjacent_swap(tape[2:], actionTranspositionChance)
    transposedTape.extend(newTapePostTransposition)

    
    newTape = []
    for idx, (actionIdentifier, actionParameters) in enumerate(transposedTape):
        if idx == 0:
            heroKey, coordinateList = actionParameters
            newCoordinateList = mutateCoordinates(coordinateList, buyCoordinateModificationChance)
            newTape.append((actionIdentifier, (heroKey, newCoordinateList)))
            continue

        elif idx == 1:
            towerKey, coordinateList = actionParameters
            newTowerKey = mutateTowerKey(towerKey, buyTowerReselectionChance)
            newCoordinateList = mutateCoordinates(coordinateList, buyCoordinateModificationChance)
            newTape.append((actionIdentifier, (newTowerKey, newCoordinateList)))
            boughtTowers[newTowerKey] += 1
            continue

        if actionIdentifier == "buy":
            if random.random() < actionTypeFlipChance:
                newUpgradeActionParameters = sampleRandomUpgrade(boughtTowers)
                newTape.append(('upgrade', newUpgradeActionParameters))
                continue

            towerKey, coordinateList = actionParameters
            newTowerKey = mutateTowerKey(towerKey, buyTowerReselectionChance)
            newCoordinateList = mutateCoordinates(coordinateList, buyCoordinateModificationChance)
            newTape.append((actionIdentifier, (newTowerKey, newCoordinateList)))
            boughtTowers[newTowerKey] += 1
            continue
                
        elif actionIdentifier == "upgrade":
            if random.random() < actionTypeFlipChance:
                towerKey, coordinates = sampleRandomPurchase()
                boughtTowers[newTowerKey] += 1
                newTape.append(('buy', (towerKey, coordinates)))
                continue

            upgradeKey, towerKey, placementIdx = actionParameters
            newUpgradeKey = mutateUpgradeKey(upgradeKey, upgradePathModificationChance)
            newTowerKey, newPlacementIdx = mutateUpgradeTowerSelectionKey(towerKey, placementIdx, boughtTowers, upgradeTowerModificationChance)
            newTape.append((actionIdentifier, (newUpgradeKey ,newTowerKey, newPlacementIdx)))
            continue
    return newTape
            
            
    
```

```python
def playUntilVictoryExclamation():
    tape = generateRandomActionTape(120)
    r, dollas, hp = rolloutTape(tape)
    time.sleep(1)
    pg.click(853, 811)
    time.sleep(1)
    pg.click(1132, 726)
    time.sleep(1)

    while True:
        newTape = modifyTape(tape)
        newR, newDollas, newHp = rolloutTape(newTape)
        if newR > r or ((newR > r) and (newDollas > dollas)):
            tape = newTape
            r = newR
            dollas = newDollas
            hp = newHp
        print(r, dollas, hp)
        time.sleep(1)
        pg.click(853, 811)
        time.sleep(1)
        pg.click(1132, 726)
        time.sleep(1)
time.sleep(3)
playUntilVictoryExclamation()
```
